```python
import os
from avac4qgis_validation import validation_case
case = validation_case('WAVE', '04_thacker_planar_paraboloid')
CORES = max(1, os.cpu_count() or 1)
case.path
```

## Physical and numerical definition

A genuinely two-dimensional planar free surface oscillates in a parabolic bowl for three periods. This tests wetting/drying and preservation of the moving analytical surface.


```python
CASE_KEY = 'thacker_planar_paraboloid'
case.run(case.path.parent / 'run_validation.py', CASE_KEY, '--solver', 'wave', '--output-root', case.path.parent, '--cores', CORES, cwd=case.path.parent)
```
