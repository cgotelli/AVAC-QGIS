# 2008 WRR dry-bottom sloping-bed benchmark — WAVE

The same frictionless water benchmark used for AVAC is executed independently with WAVE, preserving its exact wedge initial condition and quasi-one-dimensional periodic strip.


## Reproducible environment

The first code cell installs the repository's validation package and Python dependencies into the active kernel. A clean checkout also needs GNU Make and gfortran to compile the selected solver on first use.


```python
from pathlib import Path
SEARCH_ROOT = Path.cwd().resolve()
REPOSITORY = next(candidate for candidate in (SEARCH_ROOT, *SEARCH_ROOT.parents) if (candidate / 'validation' / 'pyproject.toml').is_file())
%pip install -q -e {REPOSITORY / 'validation'}
```

```python
import os
from avac4qgis_validation import validation_case
case = validation_case('WAVE', '2008_WRR_sloping_bed')
CORES = max(1, os.cpu_count() or 1)
case.path
```

## Physical and boundary conditions

The bed is a 10° slope; the one-meter triangular water wedge is bounded by a wall upstream and an extrapolating boundary downstream. Transverse boundaries are periodic.


```python
DX_M = 0.005
T_FINAL_S = 5.0
OUTPUT_FRAMES = 100
XLOWER_M = -10.0
XUPPER_M = 40.0
driver = case.path.parents[1] / 'AVAC' / '2008_WRR_sloping_bed' / 'run_avac_validation.py'
case.run(driver, '--solver', 'wave', '--output-root', case.path, '--dx', DX_M, '--t-final', T_FINAL_S, '--nout', OUTPUT_FRAMES, '--cores', CORES, '--xlower', XLOWER_M, '--xupper', XUPPER_M, '--rear-tracker', 'tutorial')
```
